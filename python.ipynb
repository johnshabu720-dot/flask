```python
def generate_fibonacci(limit):
    sequence = []
    a, b = 0, 1
    while a <= limit:
        sequence.append(a)
        a, b = b, a + b
    return sequence
limit_value = 100
fib_seq = generate_fibonacci(limit_value)
print(f"Fibonacci sequence up to {limit_value}: {fib_seq}")
```

```python
def rotate_list(lst, k):

    if not lst:
        return lst


    n = len(lst)
    k = k % n


    return lst[-k:] + lst[:-k]


original_list = [1, 2, 3, 4, 5, 6, 7]

print("Original list:  ", original_list)
print("Rotate right 2: ", rotate_list(original_list, 2))
print("Rotate left 2:  ", rotate_list(original_list, -2))
print("Rotate right 10:", rotate_list(original_list, 10))
```
